```python
import numpy as np
import scipy.integrate as spi
from scipy.interpolate import interp1d

# --- Parametri osservativi ---
T_obs = 10 * 365 * 24 * 3600  # 10 anni in secondi (modifica secondo le tue esigenze)
Omega = 4 * np.pi            # Campo visivo: 2pi sr (es. osservazione dell'emisfero)

# --- Lettura dell'area efficace ---
# Supponiamo che il file abbia:
# Colonna 0: log10(E/GeV)
# Colonna 3: Area efficace in cm²
aeff_data = np.genfromtxt('Aeff/aeff-30.txt', skip_header=1)
logE = aeff_data[:, 0]
E = 10**logE                # Energia in GeV
Aeff = aeff_data[:, 3]      # Area efficace in cm²

# Creazione della funzione di interpolazione per l'area efficace
interp_aeff = interp1d(E, Aeff, kind='linear', fill_value="extrapolate")

# --- Definizione del flusso di neutrini atmosferici ---
def phi_atm(E):
    """
    Flusso di neutrini atmosferici:
    Phi_atm(E) = 1.95e-18 * E^(-3.7) in cm⁻² s⁻¹ sr⁻¹ GeV⁻¹,
    dove E è in GeV.
    """
    return 1.95e-18 * (E)**(-3.7)

# --- Intervallo di energia di interesse (in GeV) ---
E_min = 1e2   # ad es. 100 GeV
E_max = 1e8   # ad es. 10^8 GeV (modifica in base al tuo range)

# --- Definizione dell'integranda ---
def integrand(E):
    # Phi_atm(E) [cm⁻² s⁻¹ sr⁻¹ GeV⁻¹] * Aeff(E) [cm²] * dE [GeV]
    # Il risultato è in s⁻¹ sr⁻¹
    return phi_atm(E) * interp_aeff(E)

# --- Integrazione sull'intervallo di energia ---
background_integral, err = spi.quad(integrand, E_min, E_max, limit=200)

# --- Calcolo del numero totale di eventi di background ---
# N_bkg = T_obs * Omega * (integrale)
N_bkg = T_obs * Omega * background_integral

print("Numero stimato di eventi di background:", N_bkg)
```
