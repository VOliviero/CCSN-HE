```python
import numpy as np
from scipy.integrate import quad

# Constants
c = 299792  # speed of light in km/s
H0 = 67.4   # Hubble constant in km/s/Mpc
Omega_m = 0.3  # Matter density parameter
Omega_Lambda = 0.7  # Dark energy density parameter

# Define the E(z) function
def E(z):
    return np.sqrt(Omega_m * (1 + z)**3 + Omega_Lambda)

# Function to compute the comoving distance integral
def comoving_distance(z):
    integrand = lambda z_: 1 / E(z_)
    integral, _ = quad(integrand, 0, z)
    return c * integral / H0



# Calculate the comoving distance for z = 0.002940
z_value = 0.002430 
comoving_dist_Mpc = comoving_distance(z_value)
print(comoving_dist_Mpc)
```
