```python
#This fuction divides the .dat files from Murase creating a new .dat file for each block
#La ripeto 4 volte, una per ogni file di flusso di "2N_22", "2N_20", "2P_22","2P_20"


def split_dat_file(input_file):
with open(input_file, 'r') as f:
data = f.read()

blocks = data.strip().split('\n\n') # Rimuove spazi vuoti all’inizio e alla fine e definisce un "blocco" ogni volta che i dati si interrompono da due righe vuote
for i, block in enumerate(blocks):#Itera su ogni blocco identificato, utilizzando enumerate per avere sia l'indice (i) che il contenuto (block) di ciascun blocco.
with open(f'2N_20_block_{i + 1}.dat', 'w') as f:
f.write(block) #prende il contenuto block e lo scrive in f, con nome "2N_20_block..etc"


input_file = '2N_FinalFluxs20.dat' # inserisci in input_file il file .dat originale di Murase
split_dat_file(input_file) #queste righe vanno messe alla fine perchè definisco prima tutta la funzione,
#e poi la applico
#LIBRERIA
import numpy as np
import glob #Utile per trovare file che corrispondono a uno schema specifico. Ad esempio, glob.glob('*.dat') può essere usato per elencare tutti i file .dat nella directory corrente.
import matplotlib.pyplot as plt
#leggi i file .dat con glob, elaborali con numpy, visualizza con matplotlib
#LIGHT CURVES FROM MURASE PAPER


Ottenere la lista di tutti i file .dat nella cartella SN_IIn

file_list = glob.glob('2P_20_block_*.dat')

#Vettore lightcurve e tempo per IIP S=2.0
light_curve_all_P20=[]
time_all_P20 = []

Aeff=1
#il flusso lo devi dividere per 3, perchè quello di Murase è all flavor, ipotizziamo divisione equa tra i flavor
#la formula è: (dN/dE1dt1)Aeff(E1)+(dN/dE2dt1)Aeff(E2)...+(dN/dE1dt2)Aeff(E1)+..
energy_at_time_K=[]
flux_data=[]

block_ranges = [(1, 51)]
for start_block, end_block in block_ranges:
    for file_name in file_list:
# Estrai il numero di blocco dal nome del file
    block_number = int(file_name.split('2P_20_block_')[-1].split('.')[0])
# Carica i dati dal file .dat
    new_data = np.loadtxt(file_name)

if start_block <= block_number <= end_block:

#somma su tutte le energie dei flussi, per l'i-esimo istante di tempo (k-blocco)
    column_index = 7 #la colonna di dN/dE è la ottava, partono da zero quindi è la 7
    column_data_flux = new_data [:, column_index] #estrai i dati della colonna i-esima
#moltiplico ogni flusso(Ei) per l'Aeff(Ei) corrispondente

    fluxXAeff=[]
    fluxXAeff=column_data_flux*Aeff
    #print (fluxXAeff)
    column_sum_total_fluxXAeff= np.sum(fluxXAeff) #somma degli elementi della colonna i-esima
    flux_data.append(column_sum_total_fluxXAeff)
    #print("somma",column_sum_total_fluxXAeff)
print("append", flux_data)
TOT=np.sum(flux_data)
print ("somma totale expected events",TOT)
```
